```python
missing_values = products_df.isnull().sum()

missing_pct = (missing_values/ len(products_df)) * 100

missing_df = pd.DataFrame({"missing_count": missing_values,
                          "missing_percentage": missing_pct.round(2)})

display(missing_df)
```

#### a. Checking if the missing discount columns are in the same rows where retail_price is missing

### 1.4. Checking brand, category and sub_category against product name

```python
# extract the base product name without the "(product_id)" suffix
products_df["base_name"] = (products_df["product_name"]
                             .str.replace(r"\s*\(\d+\)$", "", regex=True).str.strip())
print("Unique base products:", products_df["base_name"].nunique())
display(products_df["base_name"].value_counts())
```

```python
#the brand column agree with the product name?
brand_in_name = products_df.apply(lambda r: r["brand"].lower() in r["product_name"].lower(), axis=1)
print("Brand found in product name:", brand_in_name.sum(), "of", len(products_df))

# how many different brands / categories / sub_categories each base product is listed under
display(products_df.groupby("base_name")[["brand", "category", "sub_category"]].nunique())
```

#### Observation:
* Every product name follows the pattern `<base product> (<product_id>)` and there are only **10 base products**.
* The brand appears in the product name in only ~10% of rows, which is what random assignment across 10 brands would give.
* Each base product is listed under many different brands, categories and sub categories.

So `product_name` is the reliable field, and `brand`, `category` and `sub_category` are rebuilt from it:
* **brand**: taken from the product name itself
* **category / sub_category**: assigned using only the 3 categories and 9 sub categories that already exist in the data

#### a. Creating the product mapping

```python
product_mapping = {
    "Dell XPS 13 Intel i7 Laptop": {
        "brand": "Dell",
        "category": "Electronics",
        "sub_category": "Laptops"
    },
    "Adidas Running Shoes Ultraboost": {
        "brand": "Adidas",
        "category": "Fashion",
        "sub_category": "Footwear"
    },
    "Apple iPhone 15 (128GB) - Black": {
        "brand": "Apple",
        "category": "Electronics",
        "sub_category": "Mobiles"
    },
    "Realme Wireless Earbuds Air 5": {
        "brand": "Realme",
        "category": "Electronics",
        "sub_category": "Audio"
    },
    "Sony WH-1000XM5 Headphones": {
        "brand": "Sony",
        "category": "Electronics",
        "sub_category": "Audio"
    },
    "Puma Sports Fitness T-Shirt": {
        "brand": "Puma",
        "category": "Fashion",
        "sub_category": "Apparel"
    },
    "Cosco Tournament Football Size 5": {
        "brand": "Cosco",
        "category": "Sports",
        "sub_category": "Football"
    },
    "Nike Leather Cricket Bat Top Grade": {
        "brand": "Nike",
        "category": "Sports",
        "sub_category": "Cricket"
    },
    "Samsung Galaxy Watch 6 Smartwatch": {
        "brand": "Samsung",
        "category": "Electronics",
        "sub_category": "Wearables"
    },
    "Philips Smart LED TV 55 Inch": {
        "brand": "Philips",
        "category": "Electronics",
        "sub_category": "Televisions"
    }
}
```

#### b. Replacing brand, category and sub_category using the mapping

```python
# NEW
mapping_df = (pd.DataFrame.from_dict(product_mapping, orient="index")
                .rename_axis("base_name").reset_index())

products_df = (products_df.drop(columns=["brand", "category", "sub_category"])
                          .merge(mapping_df, on="base_name", how="left", validate="many_to_one"))

print("Unmapped products:", products_df["brand"].isna().sum())
```

### 1.5. Filling missing retail_price

Missing retail price values are filled using the median of the same base product (after the brand/category fix).

```python
# CHANGED: grouped by base_name instead of sub_category
products_df["retail_filled"] = products_df["retail_price"].isna()

group_median = products_df.groupby("base_name")["retail_price"].transform("median")
products_df["retail_price"] = products_df["retail_price"].fillna(group_median)

print("Missing retail_price after cleaning:", products_df["retail_price"].isna().sum())
```

### 1.6.Filling missing rating

Missing rating values are filled using the median of the same base product.

```python
#grouped by base_name instead of sub_category
products_df["rating_filled"] = products_df["rating"].isna()

rating_group_median = products_df.groupby("base_name")["rating"].transform("median")
products_df["rating"] = products_df["rating"].fillna(rating_group_median)

print("Missing rating after cleaning:", products_df["rating"].isna().sum())
```

### 1.7.Validating `discount_amount`, `discount_percentage` and `price_difference`

* disocunt_amount = retail_price - discount_price
* discount_percentage = discount_amount / retail_price * 100

```python
expected_amount = (products_df["retail_price"] - products_df["discount_price"]).round(2)
expected_pct = (expected_amount / products_df["retail_price"] * 100).round(2)

expected_price_difference = expected_amount.copy()

#Allowing a tolerance limit (0.02) for rounding difference, treating matching NaNs as equal
mismatch_amount = ~np.isclose(products_df["discount_amount"], expected_amount, atol=0.02, equal_nan=True)
mismatch_diff = ~np.isclose(products_df["price_difference"], expected_price_difference, atol=0.02, equal_nan=True)
mismatch_pct = ~np.isclose(products_df["discount_percentage"], expected_pct, atol=0.02, equal_nan=True)

print("Mismatched discount_amount rows:", mismatch_amount.sum())
print("Mismatched price_difference rows:", mismatch_diff.sum())
print("Mismatched discount_percentage rows:", mismatch_pct.sum())
```

```python
outside_range = ~products_df["discount_percentage"].between(0, 100)
print("discount_percentage outsied 0-100:", outside_range.sum())
```

#### a.Rebuilding `discount_price`, `discount_amount` and `price_difference`

All three columns are inconsistent with `retail_price` in almost every row, and `discount_price` is often higher than `retail_price`. `discount_percentage` is complete and within 5% – 60%, so it is kept, and the other columns are derived from it (same rule as the category catalog):

* discount_price = retail_price * (1 - discount_percentage / 100)
* discount_amount = retail_price - discount_price
* price_difference = discount_amount

```python
#rebuilt from retail_price and discount_percentage
products_df["discount_price"] = (products_df["retail_price"]
                                * (1 - products_df["discount_percentage"] / 100)).round(2)
products_df["discount_amount"] = (products_df["retail_price"] - products_df["discount_price"]).round(2)
products_df["price_difference"] = products_df["discount_amount"]

print(products_df[["discount_price", "discount_amount", "price_difference"]].isna().sum())
```

### 1.8.Filling missing rating_category

Checking what rating range for each category in the existing data and filling them based on that.

```python
products_df.groupby("rating_category")["rating"].agg(["min", "max"])
```

Rating Category Rule:
* 4.5 and above → `Excellent`
* 4.0 to 4.4 → `Good`
* < 4.0 → `Average`

```python
def get_rating_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

products_df["rating_category"] = products_df["rating"].apply(get_rating_category)

print("Missing rating_category after cleaning:", products_df["rating_category"].isnull().sum())
```

### 1.9.Duplicates

```python
print("Duplicate Rows:", products_df.duplicated().sum())
print("Duplicate product_id:", products_df["product_id"].duplicated().sum())
```

### 1.10.Check invalid prices, rating and reviews

```python
print("Retail price <= 0:", (products_df["retail_price"] <= 0).sum())
print("Discount price <=0:", (products_df["discount_price"] <= 0).sum())
print("Discount price > retail price:", (products_df["discount_price"] > products_df["retail_price"]).sum())
print("Rating outside 0-5:", ((products_df["rating"] < 0) | (products_df["rating"] > 5)).sum())
print("Negative reviews:", (products_df["num_reviews"] < 0).sum())
```

### 1.11.Veriying discount calculation and cross checking categorical columns

```python
calculated_amount = products_df["retail_price"] - products_df["discount_price"]
calculated_pct = (calculated_amount / products_df["retail_price"] ) * 100

print("discount_amount error:", ((products_df["discount_amount"] - calculated_amount).abs() > 0.05).sum())
print("discount_percentage error:", ((products_df["discount_percentage"] - calculated_pct).abs() > 0.05).sum())
print("price_difference error:", ((products_df["price_difference"] - calculated_amount).abs() > 0.05).sum())
```

```python
for col in ["category", "sub_category", "brand", "seller_name", "stock_status", "rating_category"]:
    print(f"{col}: {products_df[col].unique()}")
    print()
```

```python
#check if the rating_category matches the rating for every row
mismatch = products_df["rating"].apply(get_rating_category) != products_df["rating_category"]
print("rating_category not matching rating:", mismatch.sum())
```

```python
text_cols = ["product_name", "category", "sub_category", "brand", "seller_name", "stock_status", "rating_category"]
for col in text_cols:
    products_df[col] = products_df[col].str.strip()
```

### 1.12.Adding discount range column

```python
def get_discount_range(pct):
    if pct <= 20:
        return "0-20%"
    elif pct <= 40:
        return "21-40%"
    elif pct <= 60:
        return "41-60%"
    elif pct <= 80:
        return "61-80%"
    else:
        return "81-100%"

products_df["discount_range"] = products_df["discount_percentage"].apply(get_discount_range)
products_df["discount_range"].value_counts().sort_index()
```

```python
#drop base_name and keep schema column order
products_df = products_df.drop(columns=["retail_filled", "base_name"])

products_df = products_df[["product_id", "product_name", "category", "sub_category", "brand",
                           "retail_price", "discount_price", "discount_percentage", "discount_amount",
                           "price_difference", "rating", "num_reviews", "seller_name", "stock_status",
                           "rating_category", "discount_range", "rating_filled"]]

print("Final shape:", products_df.shape)
print(products_df.isnull().sum().sum(), "missing values left")
```

```python
#saving the file
products_df.to_csv("other_products_cleaned.csv", index=False)
print("Saved other_products_cleaned.csv")
```

# 4. Final Validation of all cleaned files

The saved CSV files are re-read and checked. Any failed check stops the notebook.

```python
def validate(df, name, priced=True):
    checks = {
        "missing values": df.isna().sum().sum(),
        "duplicate product_id": df["product_id"].duplicated().sum(),
        "rating outside 0-5": (~df["rating"].between(0, 5)).sum(),
        "negative reviews": (df["num_reviews"] < 0).sum(),
        "wrong rating_category": (df["rating"].apply(get_rating_category) != df["rating_category"]).sum(),
    }
    if priced:
        brand_in_name = df.apply(
            lambda r: str(r["brand"]).lower() in str(r["product_name"]).lower(),
            axis=1,
        )
        expected_discount = df["retail_price"] - df["discount_price"]

        checks.update({
            "brand not in product_name": (~brand_in_name).sum(),
            "discount_price > retail_price": (
                df["discount_price"] > df["retail_price"]
            ).sum(),
            "discount % outside 5-60": (
                ~df["discount_percentage"].between(5, 60)
            ).sum(),
            "discount_amount mismatch": (
                ~np.isclose(df["discount_amount"], expected_discount, atol=0.01)
            ).sum(),
            "wrong discount_range": (
                df["discount_percentage"].map(get_discount_range)
                != df["discount_range"]
            ).sum(),
        })

    report = pd.Series(checks, name=name)
    print(f"{report.to_string()}\n")
    assert report.sum() == 0, f"{name} failed validation"


datasets = [
    ("other_products_cleaned.csv", "flipkart_other_products", True),
    ("seller_performance_cleaned.csv", "seller_performance", False),
    ("category_catalog_cleaned.csv", "category_catalog", True),
]

for file_path, dataset_name, has_pricing in datasets:
    data = pd.read_csv(file_path)
    validate(data, dataset_name, priced=has_pricing)

print("All files passed validation")
```

# 6. Loading the cleaned data into MySQL

Create the database first in MySQL: `CREATE DATABASE flipkart_db;`
Required libraries: `pip install sqlalchemy pymysql`

```python
from sqlalchemy import create_engine, text

#Step 1: Connect to MySQL
server_engine = create_engine("mysql+pymysql://root:root@localhost:3306")

#create database flipkart_db
with server_engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS flipkart_db"))

#connect to the new database
engine = create_engine("mysql+pymysql://root:root@localhost:3306/flipkart_db")

#test the connection
with engine.connect() as conn:
    print("Connected to:", conn.execute(text("SELECT DATABASE()")).scalar())
```

```python
# Step 2: Write the CREATE TABLE statements
create_products = """
CREATE TABLE flipkart_other_products (
    product_id          INT PRIMARY KEY,
    product_name        VARCHAR(150),
    category            VARCHAR(50),
    sub_category        VARCHAR(50),
    brand               VARCHAR(50),
    retail_price        FLOAT,
    discount_price      FLOAT,
    discount_percentage FLOAT,
    discount_amount     FLOAT,
    price_difference    FLOAT,
    rating              FLOAT,
    num_reviews         INT,
    seller_name         VARCHAR(50),
    stock_status        VARCHAR(20),
    rating_category     VARCHAR(20),
    discount_range      VARCHAR(20),
    rating_filled       BOOLEAN
)
"""

create_catalog = """
CREATE TABLE category_catalog (
    product_id          INT PRIMARY KEY,
    product_name        VARCHAR(150),
    category            VARCHAR(50),
    sub_category        VARCHAR(50),
    brand               VARCHAR(50),
    retail_price        FLOAT,
    discount_price      FLOAT,
    discount_percentage FLOAT,
    discount_amount     FLOAT,
    price_difference    FLOAT,
    rating              FLOAT,
    num_reviews         INT,
    rating_category     VARCHAR(20),
    discount_range      VARCHAR(20),
    product_url         TEXT
)
"""

create_sellers = """
CREATE TABLE seller_performance (
    product_id      INT PRIMARY KEY,
    product_name    VARCHAR(150),
    seller_name     VARCHAR(50),
    rating          FLOAT,
    num_reviews     INT,
    stock_status    VARCHAR(20),
    rating_category VARCHAR(20)
)
"""


# Step 3: Create the tables
# (DROP first, so the cell can be run again without errors)
with engine.begin() as conn:
    conn.execute(text("DROP TABLE IF EXISTS flipkart_other_products"))
    conn.execute(text("DROP TABLE IF EXISTS category_catalog"))
    conn.execute(text("DROP TABLE IF EXISTS seller_performance"))

    conn.execute(text(create_products))
    conn.execute(text(create_catalog))
    conn.execute(text(create_sellers))

print("Tables created")
```

```python
# Step 4: Insert the cleaned data into the tables
products_df.to_sql("flipkart_other_products", engine, if_exists="append", index=False)
catalog_df.to_sql("category_catalog", engine, if_exists="append", index=False)
sellers_df.to_sql("seller_performance", engine, if_exists="append", index=False)

print("Data inserted")
```

```python
# Step 5: Verify the record counts
with engine.connect() as conn:
    products_count = conn.execute(text("SELECT COUNT(*) FROM flipkart_other_products")).scalar()
    catalog_count = conn.execute(text("SELECT COUNT(*) FROM category_catalog")).scalar()
    sellers_count = conn.execute(text("SELECT COUNT(*) FROM seller_performance")).scalar()

print("flipkart_other_products:", products_count, "rows in MySQL |", len(products_df), "rows in DataFrame")
print("category_catalog:       ", catalog_count, "rows in MySQL |", len(catalog_df), "rows in DataFrame")
print("seller_performance:     ", sellers_count, "rows in MySQL |", len(sellers_df), "rows in DataFrame")
```
